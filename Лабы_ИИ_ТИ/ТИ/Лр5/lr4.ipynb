```python
import numpy as np
import random

# Генерация стохастической матрицы

def stochastic_matrix(n):
    # Генерируем случайную матрицу
    matrix = np.random.random(size=(n, n))
    
    # Нормализуем строки
    row_sums = matrix.sum(axis=1, keepdims=True)
    stochastic_matrix = matrix / row_sums
    
    return stochastic_matrix

# Моделирование простого информационного противоборства (без влияния)
def simple_information_warfare(trust_matrix, opinion_vector, accuracy=1e-6, file_name="simple_information_warfare.txt"):
    file_write = open(file_name, 'w')

    iteration = 0
    
    current_accuracy = accuracy + 1

    while current_accuracy > accuracy:
        iteration += 1

        file_write.write(f"{"-" * 80}\n")
        file_write.write(f"Iteration: {iteration}\n")
        new_opinion_vector = np.dot(trust_matrix, opinion_vector)

        current_accuracy = 0
        for opinion_index in range(opinion_vector.__len__()):
            current_accuracy = max(current_accuracy, abs(new_opinion_vector[opinion_index] - opinion_vector[opinion_index]))

        file_write.write(f"Opinion vector: {new_opinion_vector}\n")

        opinion_vector = new_opinion_vector
    
    return opinion_vector, iteration

if __name__ == "__main__":

    # Число агентов (мнений)
    agent_count = 10

    # Точночть схождения
    accuracy = 1e-6


    # Генерация матрицы доверия
    trust_matrix = stochastic_matrix(agent_count)
    print("Матрица доверия:")
    for matrix_row in trust_matrix:
        print(matrix_row)
    print()


    # Начальный вектор мнений (случайный)
    opinion_vector = np.random.randint(1, 20, size=agent_count)
    print("Начальный вектор мнений:")
    print(opinion_vector)
    print()


    # Моделирование для игры без влияния
    print("Моделирование для игры без влияния:")
    result_opinion_vector, iterations = simple_information_warfare(trust_matrix, opinion_vector, accuracy, "no-influence.txt")
    print(f"Итераций: {iterations}")
    print("Результирующий вектор мнений:")
    print(result_opinion_vector)
    print()


    # Моделирование для игры с влиянием
    print("Моделирование для игры с влиянием:")

    # Случайный выбор агентов игроков
    size1 = 5
    size2 = 5
    
    all_indices = list(range(agent_count))
    
    # Агенты (индексы) первого игрока
    player_1_indexes = random.sample(all_indices, size1)
    remaining_indices = [i for i in all_indices if i not in player_1_indexes]
    
    # Агенты (индексы) второго игрока
    player_2_indexes = random.sample(remaining_indices, size2)

    # Мнения агентов игроков
    player_1_val = 9
    player_2_val = -10

    print(f"Агенты первого игрока: {player_1_indexes}")
    print(f"Мнение агентов первого игрока: {player_1_val}")

    print(f"Агенты второго игрока: {player_2_indexes}")
    print(f"Мнение агентов второго игрока: {player_2_val}")

    # Модификация вектора мнений
    for index in player_1_indexes:
        opinion_vector[index] = player_1_val
    for index in player_2_indexes:
        opinion_vector[index] = player_2_val

    print("Начальный вектор мнений:")
    print(opinion_vector)
    
    result_opinion_vector, iterations = simple_information_warfare(trust_matrix, opinion_vector, accuracy, "influence.txt")
    print(f"Итераций: {iterations}")
    print("Результирующий вектор мнений:")
    print(result_opinion_vector)
    print()
    
```
